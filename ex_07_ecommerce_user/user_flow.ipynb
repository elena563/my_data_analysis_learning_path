# E-commerce user flow simulator with Markov chains

```python
# user flow simple states
states = ['homepage', 'search', 'product', 'cart', 'checkout', 'exit']
terminal_states = ['exit']

# randomly generated user flow sequences
sample_sequences = [
    ['homepage', 'search', 'product', 'cart', 'checkout', 'exit'],
    ['homepage', 'search', 'product', 'exit'],
    ['homepage', 'search', 'exit'],
    ['homepage', 'exit'],
    ['homepage', 'search', 'product', 'homepage', 'exit'],
    ['product', 'cart', 'exit'],
    ['homepage', 'product', 'cart', 'checkout', 'exit'],
    ['homepage', 'search', 'product', 'exit'],
    ['homepage', 'search', 'exit'],
    ['homepage', 'exit'],
    ['homepage', 'search', 'homepage', 'exit'],
    ['product', 'cart', 'checkout', 'exit'],
    ['homepage', 'product', 'exit'],
    ['homepage', 'search', 'product', 'cart', 'exit'],
    ['product', 'exit'],
    ['homepage', 'search', 'homepage', 'exit'],
    ['homepage', 'search', 'search', 'product', 'cart', 'checkout', 'exit'],
    ['homepage', 'product', 'product', 'cart', 'exit'],
    ['homepage', 'search', 'product', 'product', 'exit'],
    ['product', 'cart', 'cart', 'checkout', 'exit'],
    ['product', 'exit'],
    ['homepage', 'search', 'product', 'cart', 'cart', 'exit'],
    ['homepage', 'product', 'cart', 'exit'],
    ['homepage', 'search', 'search', 'exit'],
    ['homepage', 'search', 'homepage', 'cart', 'checkout', 'exit'],
]
```

### Transition matrix
This matrix represents the probabilities of transitioning from one state to another in a user flow. Each row corresponds to a current state, and each column corresponds to a potential next state. The values in the matrix indicate the likelihood of moving from the current state to the next state based on observed user behavior.

#### Terminal states
Terminal states are those from which users do not transition to any other state. In this context, the 'exit' state is a terminal state, as users do not continue to any other state after exiting the website. Terminal states are important in Markov chains because they represent *absorbing states* where the process ends.

```python
transition_counts = {state: {s: 0 for s in states} for state in states}
for sequence in sample_sequences:
    for (s1, s2) in zip(sequence[:-1], sequence[1:]):
        transition_counts[s1][s2] += 1

total_transitions = {state: sum(transition_counts[state].values()) for state in states}

transition_probabilities = {state: {s: transition_counts[state][s] / total_transitions[state] if total_transitions[state] > 0 else 0 for s in states} for state in states}

# assign probability of 1 to terminal states (user exits and does not transition to any other state)
for state in terminal_states:
    transition_probabilities[state][state] = 1

transition_probabilities
```
