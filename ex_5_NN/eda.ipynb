## Data Loading and Exploration

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm

%matplotlib inline
%config InlineBackend.figure_format = 'retina'
```

The network will be trained and tested on MNIST dataset

```python
from sklearn.utils import Bunch

data = pd.read_csv('mnist_train.csv')
mnist = Bunch(
    data=data.drop(columns=['label']),
    target=data['label'],
    frame=None,
    feature_names=[f'pixel{i}' for i in range(data.drop(columns=['label']).shape[1])],
    DESCR='MNIST dataset'
)
data = mnist.data
labels = mnist.target

mnist.keys()
```

```python
n=np.random.choice(np.arange(data.shape[0]+1))
n
```

```python
test_img = data.iloc[n].values
test_label = mnist.target.iloc[n]
test_img.shape
```

```python
side_length = int(np.sqrt(test_img.shape[0]))
reshaped_test_img = test_img.reshape(side_length, side_length)
print('Image label: ' + str(test_label))
```

```python
plt.imshow(reshaped_test_img, cmap='Greys')
plt.axis('off')
plt.show()
```
